**Model used:** `llama-3.1-8b-instant` (Groq's free tier — fast, no credit card required)

## Setup: Install dependencies and configure API key

```
# Install the Groq Python SDK
!pip install -q groq
```

```
import getpass
import os
from groq import Groq

# Securely enter your free Groq API key (get one at https://console.groq.com)
os.environ["GROQ_API_KEY"] = getpass.getpass("Enter your Groq API key: ")

client = Groq(api_key=os.environ["GROQ_API_KEY"])
MODEL = "llama-3.1-8b-instant"   # free-tier, fast model. Try "llama-3.3-70b-versatile" for higher quality.

print("Groq client configured with model:", MODEL)
```

```
def generate(prompt, temperature=0.7, max_tokens=600):
    """Send a prompt to the Groq API and return the generated text."""
    response = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=temperature,
        max_tokens=max_tokens,
    )
    return response.choices[0].message.content.strip()

def show(title, text):
    print(f"\n{'='*70}\n{title}\n{'='*70}\n{text}\n")
```

## Program 1: Product Description for a Smart Fitness Watch
Design **Zero-shot**, **One-shot**, and **Few-shot** prompts to generate a product description for a Smart Fitness Watch.

```
# --- Zero-shot Prompt ---
zero_shot_1 = """Write a compelling product description for a Smart Fitness Watch."""

zero_shot_output_1 = generate(zero_shot_1)
show("ZERO-SHOT OUTPUT", zero_shot_output_1)
```

```
# --- One-shot Prompt ---
one_shot_1 = """Write a compelling product description for a product.

Example:
Product: Wireless Bluetooth Earbuds
Description: Experience true wireless freedom with our Bluetooth Earbuds - featuring crystal-clear sound, 30-hour battery life, and a sweat-resistant design built for every workout. Compact, comfortable, and always ready to keep up with your day.

Now write a similar description for:
Product: Smart Fitness Watch"""

one_shot_output_1 = generate(one_shot_1)
show("ONE-SHOT OUTPUT", one_shot_output_1)
```

```
# --- Few-shot Prompt ---
few_shot_1 = """Write a compelling product description for a product.

Example 1:
Product: Wireless Bluetooth Earbuds
Description: Experience true wireless freedom with our Bluetooth Earbuds - featuring crystal-clear sound, 30-hour battery life, and a sweat-resistant design built for every workout. Compact, comfortable, and always ready to keep up with your day.

Example 2:
Product: Portable Power Bank
Description: Never run out of power again. Our 20,000mAh Portable Power Bank delivers fast charging for all your devices, wrapped in a sleek, pocket-friendly design perfect for travel, work, or everyday emergencies.

Now write a similar description for:
Product: Smart Fitness Watch"""

few_shot_output_1 = generate(few_shot_1)
show("FEW-SHOT OUTPUT", few_shot_output_1)
```

**Observation:** Zero-shot relies purely on the model's general knowledge, while one-shot and few-shot prompts guide the tone, structure, and length by providing example(s), typically producing more consistent and on-brand marketing copy as more examples are added.

## Program 2: 200-word Blog on "Applications of Artificial Intelligence in Healthcare"
Design **Zero-shot**, **One-shot**, and **Few-shot** prompts to generate a 200-word blog.

```
# --- Zero-shot Prompt ---
zero_shot_2 = """Write a 200-word blog post on the topic: "Applications of Artificial Intelligence in Healthcare"."""

zero_shot_output_2 = generate(zero_shot_2)
show("ZERO-SHOT OUTPUT", zero_shot_output_2)
print("Word count:", len(zero_shot_output_2.split()))
```

```
# --- Few-shot Prompt ---
few_shot_2 = """Write a 200-word blog post on a given topic, following the style below.

Example 1:
Topic: "The Rise of Renewable Energy"
Blog: Renewable energy is reshaping how the world powers itself. Solar, wind, and hydro sources are becoming cheaper and more efficient every year, reducing dependence on fossil fuels. Countries are investing heavily in clean energy infrastructure, driven by both environmental urgency and economic opportunity. Technologies like battery storage are solving the intermittency problem, making renewables more reliable than ever. Beyond the environment, this shift is creating millions of jobs and driving innovation across industries. As costs continue to fall, renewable energy is no longer a niche choice - it is becoming the default.

Example 2:
Topic: "The Growth of Remote Work"
Blog: Remote work has moved from a temporary necessity to a permanent feature of the modern workplace. Advances in communication tools, cloud computing, and project management platforms have made it possible for teams to collaborate effectively from anywhere. Employees increasingly value flexibility, and companies are recognizing that productivity does not depend on physical presence. This shift has also expanded hiring pools beyond geographic limits, allowing businesses to access global talent. However, remote work brings challenges too, including maintaining company culture and preventing burnout. Organizations that invest in the right tools and policies are best positioned to thrive in this new era of distributed work.

Now write a similar 200-word blog post on:
Topic: "Applications of Artificial Intelligence in Healthcare"
"""

few_shot_output_2 = generate(few_shot_2)
show("FEW-SHOT OUTPUT", few_shot_output_2)
print("Word count:", len(few_shot_output_2.split()))
```

**Observation:** The few-shot prompt gives the model two reference blogs to imitate in tone, paragraph flow, and length, generally producing outputs closest to the requested 200-word target.

## Program 3: Summarize an Article into 50 Words
Design **Zero-shot**, **One-shot**, and **Few-shot** prompts to summarize an article, then compare the summaries on **accuracy, completeness, and readability**.

```
# The article to summarize (replace with any article of your choice)
article = """
Artificial Intelligence (AI) is rapidly transforming industries across the globe. In healthcare, AI-powered
systems assist doctors in diagnosing diseases earlier and more accurately by analyzing medical images and
patient data. In finance, AI algorithms detect fraudulent transactions in real time and provide personalized
investment advice. Retail businesses use AI to predict customer preferences and optimize inventory management.
Meanwhile, in education, AI-driven platforms adapt learning material to each student's pace and needs. Despite
these benefits, the rise of AI also raises concerns about job displacement, data privacy, and the need for
ethical guidelines to govern its use. Governments and organizations worldwide are now working to create
regulatory frameworks that balance innovation with responsible AI deployment, ensuring that the technology
benefits society while minimizing potential risks.
"""
print(article)
```

```
# --- Zero-shot Prompt ---
zero_shot_3 = f"""Summarize the following article in exactly 50 words:

{article}"""

zero_shot_output_3 = generate(zero_shot_3, temperature=0.3)
show("ZERO-SHOT SUMMARY", zero_shot_output_3)
print("Word count:", len(zero_shot_output_3.split()))
```

```
# --- One-shot Prompt ---
one_shot_3 = f"""Summarize an article in exactly 50 words.

Example:
Article: "Climate change is accelerating due to rising greenhouse gas emissions, primarily from fossil fuels,
deforestation, and industrial activity. Rising global temperatures are causing more frequent extreme weather
events, melting polar ice, and rising sea levels, threatening ecosystems and human communities. Scientists urge
immediate action through renewable energy adoption, carbon reduction policies, and international cooperation to
limit long-term damage and protect future generations from irreversible environmental consequences worldwide."
Summary: Climate change, driven by greenhouse gas emissions from fossil fuels and deforestation, is causing extreme
weather, melting ice, and rising seas. Scientists stress urgent action via renewable energy, carbon reduction, and
global cooperation to prevent irreversible environmental damage and protect future generations.

Now summarize this article in exactly 50 words:
{article}"""

one_shot_output_3 = generate(one_shot_3, temperature=0.3)
show("ONE-SHOT SUMMARY", one_shot_output_3)
print("Word count:", len(one_shot_output_3.split()))
```

```
# --- Few-shot Prompt ---
few_shot_3 = f"""Summarize an article in exactly 50 words.

Example 1:
Article: "Climate change is accelerating due to rising greenhouse gas emissions, primarily from fossil fuels,
deforestation, and industrial activity. Rising global temperatures are causing more frequent extreme weather
events, melting polar ice, and rising sea levels, threatening ecosystems and human communities. Scientists urge
immediate action through renewable energy adoption, carbon reduction policies, and international cooperation to
limit long-term damage and protect future generations from irreversible environmental consequences worldwide."
Summary: Climate change, driven by greenhouse gas emissions from fossil fuels and deforestation, is causing extreme
weather, melting ice, and rising seas. Scientists stress urgent action via renewable energy, carbon reduction, and
global cooperation to prevent irreversible environmental damage and protect future generations.

Example 2:
Article: "Remote work has become a lasting feature of the modern workplace, driven by advances in cloud
computing and communication tools. Employees value flexibility, and companies benefit from access to global
talent pools. However, challenges remain, including maintaining company culture, ensuring productivity, and
preventing employee burnout. Organizations that invest in effective remote work policies and technology are
better positioned to succeed in this evolving, distributed work environment moving forward into the future."
Summary: Remote work is now permanent, enabled by cloud computing and communication tools, offering flexibility and
global talent access. Companies face challenges maintaining culture and preventing burnout. Success depends on
strong remote-work policies and technology investment as distributed work continues evolving.

Now summarize this article in exactly 50 words:
{article}"""

few_shot_output_3 = generate(few_shot_3, temperature=0.3)
show("FEW-SHOT SUMMARY", few_shot_output_3)
print("Word count:", len(few_shot_output_3.split()))
```

## Program 4: Professional Leave Email due to Illness
Design **Zero-shot**, **One-shot**, and **Few-shot** prompts to generate a professional email requesting leave due to illness, then compare on **tone, grammar, formatting, and completeness**.

```
# --- Zero-shot Prompt ---
zero_shot_4 = """Write a professional email to my manager requesting leave for 2 days due to illness."""

zero_shot_output_4 = generate(zero_shot_4)
show("ZERO-SHOT EMAIL", zero_shot_output_4)
```

```
# --- One-shot Prompt ---
one_shot_4 = """Write a professional email based on the example and situation given.

Example:
Situation: Requesting leave for a family emergency.
Email:
Subject: Leave Request - Family Emergency

Dear [Manager's Name],

I am writing to inform you that I need to take leave from [start date] to [end date] due to a family emergency
that requires my immediate attention. I will ensure that my ongoing tasks are handed over appropriately and will
remain reachable by email for anything urgent.

Please let me know if any additional documentation is required. Thank you for your understanding.

Best regards,
[Your Name]

Now write a similar professional email for:
Situation: Requesting 2 days of leave due to illness."""

one_shot_output_4 = generate(one_shot_4)
show("ONE-SHOT EMAIL", one_shot_output_4)
```

```
# --- Few-shot Prompt ---
few_shot_4 = """Write a professional email based on the examples and situation given.

Example 1:
Situation: Requesting leave for a family emergency.
Email:
Subject: Leave Request - Family Emergency

Dear [Manager's Name],

I am writing to inform you that I need to take leave from [start date] to [end date] due to a family emergency
that requires my immediate attention. I will ensure that my ongoing tasks are handed over appropriately and will
remain reachable by email for anything urgent.

Please let me know if any additional documentation is required. Thank you for your understanding.

Best regards,
[Your Name]

Example 2:
Situation: Requesting leave to attend a relative's wedding.
Email:
Subject: Leave Request - Family Wedding

Dear [Manager's Name],

I would like to request leave from [start date] to [end date] to attend my relative's wedding. I will complete
my pending tasks before I leave and coordinate with the team to ensure a smooth handover during my absence.

Please let me know if you need any further information. Thank you for considering my request.

Best regards,
[Your Name]

Now write a similar professional email for:
Situation: Requesting 2 days of leave due to illness."""

few_shot_output_4 = generate(few_shot_4)
show("FEW-SHOT EMAIL", few_shot_output_4)
```

**Manual Comparison (fill in after reviewing the outputs above):**

| Criteria | Zero-shot | One-shot | Few-shot |
|---|---|---|---|
| **Tone** (polite, professional) | | | |
| **Grammar** (correctness, sentence structure) | | | |
| **Formatting** (subject line, greeting, sign-off, paragraphing) | | | |
| **Completeness** (dates, reason, handover mention, closing) | | | |

*Few-shot prompts usually yield the most consistently formatted email - including a clear subject line, structured body, and professional sign-off - because the examples explicitly demonstrate the expected format.*
